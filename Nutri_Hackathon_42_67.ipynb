```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/Train_Data.csv')
df
```

```python
df.fillna(df.median(numeric_only=True), inplace=True)
df.isnull().sum()
```

```python
test_data = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/Test_Data.csv')
test_data
```

```python
df['target']= df.age_group.apply(lambda x: 1 if x == 'Senior' else 0)
df
```

```python
from sklearn.model_selection import train_test_split
#from sklearn.model_selection import KFold
#kf = KFold(n_splits = 10)

X= df.drop(['target','age_group','SEQN'],axis='columns')
y= df.target

X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score
from sklearn.svm import SVC

#model = RandomForestClassifier(n_estimators=200)
model = SVC(class_weight='balanced',kernel='linear')
model.get_params()
```

```python
#cross_val_score(RandomForestClassifier(n_estimators=200),X,y,cv=5)
```

```python
#cross_val_score(LogisticRegression(),X,y,cv=5)
```

```python
model.fit(X_train,y_train)
model.score(X_test,y_test)
```

```python
test_data.fillna(test_data.interpolate(), inplace=True)
X_test_data = test_data[['RIAGENDR','PAQ605','BMXBMI','LBXGLU','DIQ010','LBXGLT','LBXIN']]
y_predicted = model.predict(X_test_data)
```
